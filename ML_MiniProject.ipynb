```python
import os
import json
import xml.etree.ElementTree as ET
from collections import Counter

dataset_path = "/content/roadsight_dataset"
image_folder = os.path.join(dataset_path, "annotated-images")

with open(os.path.join(dataset_path, "splits.json"), "r") as f:
    splits = json.load(f)

print("========== SPLIT COUNTS ==========")
for split_name, annotation_names in splits.items():
    print(f"{split_name}: {len(annotation_names)} annotations")

print("\n========== DATA QUALITY CHECK ==========")

all_labels = Counter()
missing_images = []
missing_annotations = []
invalid_boxes = []
empty_annotations = []
split_images = {}
total_boxes = 0

for split_name, annotation_names in splits.items():
    split_images[split_name] = set()

    for annotation_name in annotation_names:
        xml_path = os.path.join(image_folder, annotation_name)

        if not os.path.exists(xml_path):
            missing_annotations.append(annotation_name)
            continue

        try:
            root = ET.parse(xml_path).getroot()
        except ET.ParseError:
            print("Unreadable XML:", annotation_name)
            continue

        image_name = root.findtext("filename")

        if not image_name:
            image_name = os.path.splitext(annotation_name)[0] + ".jpg"

        image_path = os.path.join(image_folder, image_name)

        if not os.path.exists(image_path):
            missing_images.append(image_name)
        else:
            split_images[split_name].add(image_name)

        objects = root.findall("object")

        if not objects:
            empty_annotations.append(annotation_name)

        width = int(root.findtext("size/width", "0"))
        height = int(root.findtext("size/height", "0"))

        for obj in objects:
            label = obj.findtext("name", "unknown").strip().lower()
            all_labels[label] += 1
            total_boxes += 1

            box = obj.find("bndbox")
            if box is None:
                invalid_boxes.append(annotation_name)
                continue

            try:
                xmin = float(box.findtext("xmin"))
                ymin = float(box.findtext("ymin"))
                xmax = float(box.findtext("xmax"))
                ymax = float(box.findtext("ymax"))

                if not (
                    0 <= xmin < xmax <= width
                    and 0 <= ymin < ymax <= height
                ):
                    invalid_boxes.append(annotation_name)
            except (TypeError, ValueError):
                invalid_boxes.append(annotation_name)

print("\nObject labels:", dict(all_labels))
print("Total annotated objects:", total_boxes)
print("Missing XML files:", len(missing_annotations))
print("Missing images:", len(missing_images))
print("Invalid bounding boxes:", len(invalid_boxes))
print("Images with no annotated objects:", len(empty_annotations))

print("\n========== SPLIT OVERLAP CHECK ==========")
split_names = list(split_images.keys())

for i in range(len(split_names)):
    for j in range(i + 1, len(split_names)):
        a, b = split_names[i], split_names[j]
        overlap = split_images[a] & split_images[b]
        print(f"{a} vs {b}: {len(overlap)} shared images")

print("\nValidation check complete.")
```

```python
import os
import json
import random
import shutil
import xml.etree.ElementTree as ET
from collections import Counter

# Paths
source_folder = "/content/roadsight_dataset/annotated-images"
split_file = "/content/roadsight_dataset/splits.json"
output_folder = "/content/roadsight_yolo"

# Read existing train/test split
with open(split_file, "r") as f:
    splits = json.load(f)

# Create validation split from training data
random.seed(42)
train_annotations = list(splits["train"])
random.shuffle(train_annotations)

val_count = round(len(train_annotations) * 0.20)
val_annotations = train_annotations[:val_count]
train_annotations = train_annotations[val_count:]
test_annotations = list(splits["test"])

print("New split sizes:")
print("Train:", len(train_annotations))
print("Validation:", len(val_annotations))
print("Test:", len(test_annotations))

# Create YOLO directory structure
for split_name in ["train", "val", "test"]:
    os.makedirs(f"{output_folder}/images/{split_name}", exist_ok=True)
    os.makedirs(f"{output_folder}/labels/{split_name}", exist_ok=True)

def convert_annotation(xml_name, split_name):
    xml_path = os.path.join(source_folder, xml_name)

    root = ET.parse(xml_path).getroot()
    image_name = root.findtext("filename")

    if not image_name:
        image_name = os.path.splitext(xml_name)[0] + ".jpg"

    image_path = os.path.join(source_folder, image_name)

    if not os.path.isfile(image_path):
        return False, 0, 0

    width = int(root.findtext("size/width"))
    height = int(root.findtext("size/height"))

    yolo_lines = []
    invalid_count = 0

    for obj in root.findall("object"):
        box = obj.find("bndbox")

        if box is None:
            invalid_count += 1
            continue

        try:
            xmin = float(box.findtext("xmin"))
            ymin = float(box.findtext("ymin"))
            xmax = float(box.findtext("xmax"))
            ymax = float(box.findtext("ymax"))
        except (TypeError, ValueError):
            invalid_count += 1
            continue

        # Reject invalid boxes; do not silently alter coordinates
        if not (
            0 <= xmin < xmax <= width
            and 0 <= ymin < ymax <= height
        ):
            invalid_count += 1
            continue

        # YOLO format: class_id, center_x, center_y, width, height
        x_center = ((xmin + xmax) / 2) / width
        y_center = ((ymin + ymax) / 2) / height
        box_width = (xmax - xmin) / width
        box_height = (ymax - ymin) / height

        yolo_lines.append(
            f"0 {x_center:.6f} {y_center:.6f} "
            f"{box_width:.6f} {box_height:.6f}"
        )

    # Copy image and save corresponding label file
    destination_image = (
        f"{output_folder}/images/{split_name}/{image_name}"
    )
    shutil.copy2(image_path, destination_image)

    label_name = os.path.splitext(image_name)[0] + ".txt"
    label_path = f"{output_folder}/labels/{split_name}/{label_name}"

    with open(label_path, "w") as f:
        f.write("\n".join(yolo_lines))

    return True, len(yolo_lines), invalid_count

# Convert all three splits
stats = Counter()
failed_images = []

for split_name, annotation_list in [
    ("train", train_annotations),
    ("val", val_annotations),
    ("test", test_annotations)
]:
    for xml_name in annotation_list:
        try:
            success, boxes, invalid = convert_annotation(
                xml_name, split_name
            )

            if success:
                stats[f"{split_name}_images"] += 1
                stats[f"{split_name}_boxes"] += boxes
                stats["invalid_boxes_skipped"] += invalid
            else:
                failed_images.append(xml_name)

        except Exception as e:
            failed_images.append(f"{xml_name}: {e}")

# Create dataset configuration for YOLO
yaml_content = """path: /content/roadsight_yolo
train: images/train
val: images/val
test: images/test
names:
  0: pothole
"""

with open(f"{output_folder}/data.yaml", "w") as f:
    f.write(yaml_content)

print("\n========== CONVERSION SUMMARY ==========")
for key, value in stats.items():
    print(f"{key}: {value}")

print("Failed images:", len(failed_images))
print("Invalid bounding boxes skipped:", stats["invalid_boxes_skipped"])
print("YOLO configuration created:", f"{output_folder}/data.yaml")

if failed_images:
    print("\nFirst failed items:", failed_images[:10])

print("\nConversion finished.")
```

```python
import os
import json
import random
import shutil
import xml.etree.ElementTree as ET
from collections import Counter

source_folder = "/content/roadsight_dataset/annotated-images"
split_file = "/content/roadsight_dataset/splits.json"
output_folder = "/content/roadsight_yolo"

# Check source files
assert os.path.exists(split_file), "splits.json not found"
assert os.path.isdir(source_folder), "Image folder not found"

with open(split_file, "r") as f:
    splits = json.load(f)

random.seed(42)
train_list = list(splits["train"])
random.shuffle(train_list)

val_count = round(len(train_list) * 0.20)
val_list = train_list[:val_count]
train_list = train_list[val_count:]
test_list = list(splits["test"])

# Rebuild the output folders
if os.path.exists(output_folder):
    shutil.rmtree(output_folder)

for split in ["train", "val", "test"]:
    os.makedirs(f"{output_folder}/images/{split}", exist_ok=True)
    os.makedirs(f"{output_folder}/labels/{split}", exist_ok=True)

stats = Counter()
failed = []

def convert_xml(xml_name, split_name):
    xml_path = os.path.join(source_folder, xml_name)
    root = ET.parse(xml_path).getroot()

    image_name = root.findtext("filename")
    if not image_name:
        image_name = os.path.splitext(xml_name)[0] + ".jpg"

    image_path = os.path.join(source_folder, image_name)
    if not os.path.isfile(image_path):
        raise FileNotFoundError(image_path)

    width = int(root.findtext("size/width"))
    height = int(root.findtext("size/height"))
    lines = []
    skipped = 0

    for obj in root.findall("object"):
        box = obj.find("bndbox")
        if box is None:
            skipped += 1
            continue

        try:
            x1 = float(box.findtext("xmin"))
            y1 = float(box.findtext("ymin"))
            x2 = float(box.findtext("xmax"))
            y2 = float(box.findtext("ymax"))
        except (TypeError, ValueError):
            skipped += 1
            continue

        if not (0 <= x1 < x2 <= width and 0 <= y1 < y2 <= height):
            skipped += 1
            continue

        xc = ((x1 + x2) / 2) / width
        yc = ((y1 + y2) / 2) / height
        bw = (x2 - x1) / width
        bh = (y2 - y1) / height

        lines.append(f"0 {xc:.6f} {yc:.6f} {bw:.6f} {bh:.6f}")

    shutil.copy2(
        image_path,
        f"{output_folder}/images/{split_name}/{image_name}"
    )

    label_name = os.path.splitext(image_name)[0] + ".txt"
    with open(f"{output_folder}/labels/{split_name}/{label_name}", "w") as f:
        f.write("\n".join(lines))

    return len(lines), skipped

for split_name, items in [
    ("train", train_list),
    ("val", val_list),
    ("test", test_list)
]:
    for xml_name in items:
        try:
            boxes, skipped = convert_xml(xml_name, split_name)
            stats[f"{split_name}_images"] += 1
            stats[f"{split_name}_boxes"] += boxes
            stats["invalid_boxes_skipped"] += skipped
        except Exception as e:
            failed.append(f"{xml_name}: {e}")

yaml_content = f"""path: {output_folder}
train: images/train
val: images/val
test: images/test
names:
  0: pothole
"""

with open(f"{output_folder}/data.yaml", "w") as f:
    f.write(yaml_content)

print("CONVERSION COMPLETE")
print(dict(stats))
print("Failed files:", len(failed))
print("Config exists:", os.path.exists(f"{output_folder}/data.yaml"))

if failed:
    print("First errors:", failed[:5])
```
